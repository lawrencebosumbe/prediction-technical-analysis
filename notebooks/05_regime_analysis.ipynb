# Regime analysis
Inspect trend-strength regimes using the ADX indicator.

```python
from tpta.data import load_prices
from tpta.indicators.trend import adx
prices = load_prices('US500')
prices['adx'] = adx(prices)
prices['regime'] = prices.adx.map(lambda x: 'weak' if x < 20 else 'strong')
prices[['adx', 'regime']].tail(10)
```
