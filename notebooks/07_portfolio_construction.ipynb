# Portfolio construction
Combine sample returns and inspect diversification. Missing bars are treated as zero return in this simple illustration.

```python
import pandas as pd
from tpta.data import load_prices
from tpta.backtest.portfolio import combine_returns
symbols = ['US500', 'US100', 'DAX30', 'BTCUSD']
returns = {s: load_prices(s).close.pct_change().dropna() for s in symbols}
portfolio = combine_returns(returns)
portfolio.describe()
```

```python
pd.concat(returns, axis=1).corr().style.background_gradient(cmap='coolwarm', vmin=-1, vmax=1)
```
