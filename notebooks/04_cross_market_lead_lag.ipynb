# Cross-market return relationships
Correlations are descriptive and do not establish causal lead-lag effects.

```python
import pandas as pd
from tpta.data import load_prices
symbols = ['US30', 'US500', 'US100', 'DAX30', 'NIKKEI225']
prices = pd.concat({s: load_prices(s).close for s in symbols}, axis=1)
returns = prices.pct_change()
returns.corr().round(2)
```

```python
returns.corr().style.background_gradient(cmap='coolwarm', vmin=-1, vmax=1)
```
