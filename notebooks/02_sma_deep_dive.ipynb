# SMA deep dive
Compare short and long moving averages and their directional state.

```python
from tpta.data import load_prices
from tpta.indicators import sma
from tpta.signals import sma_crossover
prices = load_prices('US500')
prices['sma_fast'] = sma(prices.close, 5)
prices['sma_slow'] = sma(prices.close, 12)
prices[['close', 'sma_fast', 'sma_slow']].plot(figsize=(10, 4))
signal = sma_crossover(prices, 'sma_fast', 'sma_slow')
signal.value_counts()
```
