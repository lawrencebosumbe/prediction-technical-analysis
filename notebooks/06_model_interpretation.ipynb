# Model interpretation
Train a small baseline and inspect feature coefficients. The sample is too short for strategy conclusions.

```python
from tpta.data import load_prices
from tpta.model.features import make_features
from tpta.model.targets import forward_return_target
from tpta.model.trainers import LogisticRegressionTrainer
from tpta.model.explain import feature_importance
prices = load_prices('BTCUSD')
features = make_features(prices)
target = forward_return_target(prices.close, horizon=2)
model = LogisticRegressionTrainer().fit(features, target)
feature_importance(model)
```
